#### 1. Write a conditional that prints different messages if a bank account balance is below 3,000, between 3,000 and 10,000, or over 10,000.

```python
# Your code goes here
def bank_account(value):
  if value < 3000:
    print("Funds below $3,000")
  elif value < 10000:
    print("Funds between $3,000 and $10,000")
  else:
    print("Funds exceed $10,000")
```

#### 3. Given the list `people`, sort it by people's first name, last name and age. Store the sorted lists as by_first_name, by_last_name, and by_age, respectively.

people = [('Mark', 'Harrison', 56), ('Ken', 'Wolseley', 23), ('Emily', 'Robinson', 77)]

```python
# Create the list
people = [('Mark', 'Harrison', 56), ('Ken', 'Wolseley', 23), ('Emily', 'Robinson', 77)]

# Sort by first name
by_first_name = sorted(people)
by_first_name
```

```python
# Notice that sorted() orders by the first element in the tuple.

# Sort by last name
# Create a new list with new tuples; each with the last name as first element.
new_list = []
for i in people:
  tup = (i[1], i[0], i[2])
  new_list.append(tup)

# Sort by first element (which is now the last name)
by_last_name = sorted(new_list)
by_last_name
```

```python
# Sort by age
# Create a new list with new tuples; each with the age as first element.
new_list = []
for i in people:
  tup = (i[2], i[0], i[1])
  new_list.append(tup)

# Sort by first element (which is now the age)
by_last_name = sorted(new_list)
by_last_name
```

#### 4. Write a function called `dict_intersect` that takes two dictionaries, d1 and d2, as arguments and returns a set that contains only the keys found in both of the original dictionaries.

```python
# Create function
def dict_intersect(d1, d2):
  keys1 = set(d1.keys())
  keys2 = set(d2.keys())
  return keys1.intersection(keys2)
```

```python
# Check
d1 = {"Germany": "Berlin", "Italy": "Rome", "France": "Paris"}
d2 = {"Spain": "Madrid", "Italy": "Rome", "France": "Paris"}

dict_intersect(d1, d2)
```
