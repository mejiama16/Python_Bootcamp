#### 1. Create a pandas DataFrame named `df` using the following table of names, ages, and scores.

| Anna  | 23  | 88    |

```python
# Your code goes here
df = pd.DataFrame({
    "Name": ["Anna", "Ben", "Cara", "David"],
    "Age": [23, 35, 29, 41],
    "Score": [88, 92, 79, 85]
})
```

## Matplotlib

```python
import matplotlib.pyplot as plt
```

#### 1. **Bar Plot.** Create a bar chart showing each person’s Name on the x-axis and their Score on the y-axis.

```python
# Bar chart
plt.bar(df["Name"], df["Score"], color='skyblue', edgecolor='navy')

# Add labels
plt.xlabel('Name')
plt.ylabel('Score')
plt.title('Names and Scores of Students')

plt.show()
```

#### 2. **Scatter Plot.** Create a scatter plot for df showing the relationship between `Age ` and `Score`.

```python
# Scatterplot
plt.scatter(df["Age"], df["Score"])

# Add labels
plt.xlabel('Age')
plt.ylabel('Score')
plt.title('Ages and Scores of Students')

plt.show()
```

#### 3. **Histogram**. Create a histogram to show the distribution of the `Score` column.

```python
# Your code goes here
plt.hist(df["Score"], color='skyblue', edgecolor='navy')

# Add labels
plt.xlabel('Score')
plt.ylabel('Count')
plt.title('Histogram of Scores Distribution')

plt.show()
```
